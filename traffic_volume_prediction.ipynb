```python
import numpy as np
```

```python
def circular_mean(degrees):
    radians = np.deg2rad(degrees.dropna())
    
    if len(radians) == 0:
        return np.nan
    
    sin_mean = np.mean(np.sin(radians))
    cos_mean = np.mean(np.cos(radians))
    
    mean_angle = np.arctan2(sin_mean, cos_mean)
    mean_angle = np.rad2deg(mean_angle)
    
    return mean_angle % 360
```

```python
wind_direction_clean = (
    df.groupby("date_time")["wind_direction"]
      .apply(circular_mean)
      .reset_index()
)
```

```python
# Create the complete expected hourly timeline
full_timeline = pd.date_range(
    start=df_clean["date_time"].min(),
    end=df_clean["date_time"].max(),
    freq="h"
)

# Identify missing timestamps
observed_timestamps = pd.DatetimeIndex(df_clean["date_time"])

missing_timestamps = full_timeline.difference(observed_timestamps)

print("Expected hourly timestamps:", len(full_timeline))
print("Observed timestamps:", len(observed_timestamps))
print("Missing timestamps:", len(missing_timestamps))

print(
    "\nMissing percentage:",
    round((len(missing_timestamps) / len(full_timeline)) * 100, 2),
    "%"
)

print("\nFirst 20 missing timestamps:")
print(missing_timestamps[:20])
```

```python
# Convert missing timestamps into a DataFrame
missing_df = pd.DataFrame({"date_time": missing_timestamps})

# Calculate time difference between consecutive missing timestamps
missing_df["time_diff"] = missing_df["date_time"].diff()

# Identify new missing blocks
missing_df["new_block"] = (
    missing_df["time_diff"] != pd.Timedelta(hours=1)
)

# Assign block IDs
missing_df["block_id"] = missing_df["new_block"].cumsum()

# Summarize missing blocks
gap_summary = (
    missing_df.groupby("block_id")
    .agg(
        start=("date_time", "min"),
        end=("date_time", "max"),
        missing_hours=("date_time", "count")
    )
)

gap_summary["duration_hours"] = (
    (gap_summary["end"] - gap_summary["start"])
    .dt.total_seconds() / 3600 + 1
)

# Sort by largest missing blocks
gap_summary = gap_summary.sort_values(
    "missing_hours",
    ascending=False
)

print("Number of missing blocks:", len(gap_summary))

print("\nLargest missing blocks:")
print(gap_summary.head(15))
```

```python
# Create chronological train-validation-test split

train_df = df[df["year"] <= 2015].copy()

val_df = df[df["year"] == 2016].copy()

test_df = df[df["year"] == 2017].copy()


print("Training Set:", train_df.shape)
print("Validation Set:", val_df.shape)
print("Test Set:", test_df.shape)


print("\nDate Ranges:")

print(
    "Train:",
    train_df["date_time"].min(),
    "to",
    train_df["date_time"].max()
)

print(
    "Validation:",
    val_df["date_time"].min(),
    "to",
    val_df["date_time"].max()
)

print(
    "Test:",
    test_df["date_time"].min(),
    "to",
    test_df["date_time"].max()
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Temporal features
temporal_features = [
    "hour",
    "day_of_week",
    "month",
    "is_weekend",
    "is_holiday_flag"
]

target = "traffic_volume"

# Prepare datasets
X_train = train_df[temporal_features]
y_train = train_df[target]

X_val = val_df[temporal_features]
y_val = val_df[target]

# Train Linear Regression baseline
lr_model = LinearRegression()

lr_model.fit(X_train, y_train)

# Predictions
val_predictions = lr_model.predict(X_val)

# Evaluation
mae = mean_absolute_error(y_val, val_predictions)
rmse = np.sqrt(mean_squared_error(y_val, val_predictions))
r2 = r2_score(y_val, val_predictions)

print("===== Linear Regression Temporal Baseline =====")
print(f"MAE:  {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²:   {r2:.4f}")
```

```python
from sklearn.ensemble import RandomForestRegressor

# Random Forest Temporal Model
rf_temporal = RandomForestRegressor(
    n_estimators=200,
    max_depth=20,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

# Train model
rf_temporal.fit(X_train, y_train)

# Validation predictions
rf_val_predictions = rf_temporal.predict(X_val)

# Evaluation
rf_mae = mean_absolute_error(y_val, rf_val_predictions)
rf_rmse = np.sqrt(mean_squared_error(y_val, rf_val_predictions))
rf_r2 = r2_score(y_val, rf_val_predictions)

print("===== Random Forest Temporal Model =====")
print(f"MAE:  {rf_mae:.2f}")
print(f"RMSE: {rf_rmse:.2f}")
print(f"R²:   {rf_r2:.4f}")
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor

configs = [
    {
        "learning_rate": 0.03,
        "max_iter": 700,
        "max_leaf_nodes": 15,
        "l2_regularization": 1.0
    },
    {
        "learning_rate": 0.05,
        "max_iter": 700,
        "max_leaf_nodes": 31,
        "l2_regularization": 1.0
    },
    {
        "learning_rate": 0.05,
        "max_iter": 500,
        "max_leaf_nodes": 63,
        "l2_regularization": 1.0
    },
    {
        "learning_rate": 0.08,
        "max_iter": 500,
        "max_leaf_nodes": 31,
        "l2_regularization": 2.0
    }
]

results = []

for i, config in enumerate(configs, start=1):

    model = HistGradientBoostingRegressor(
        **config,
        random_state=42
    )

    model.fit(X_train_core, y_train)

    predictions = model.predict(X_val_core)

    mae = mean_absolute_error(y_val, predictions)
    rmse = np.sqrt(mean_squared_error(y_val, predictions))
    r2 = r2_score(y_val, predictions)

    results.append({
        "Model": f"HGB_{i}",
        **config,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    "RMSE"
).reset_index(drop=True)

print(results_df)
```

```python
# Combine Train and Validation datasets
X_train_final = pd.concat([
    X_train_core,
    X_val_core
])

y_train_final = pd.concat([
    y_train,
    y_val
])

# Test features and target
X_test_final = test_df[core_features]
y_test_final = test_df["traffic_volume"]

print("Final Training Set:", X_train_final.shape)
print("Final Test Set:", X_test_final.shape)

print("\nTraining period:")
print("Train + Validation: 2012-10-02 to 2016-12-31")

print("\nTest period:")
print("2017-01-01 to 2017-05-17")
```

```python
# Create prediction results dataframe

results_df = test_df[['date_time', 'hour', 'day_of_week', 'is_weekend']].copy()

results_df['actual_traffic'] = y_test_final.values
results_df['predicted_traffic'] = y_pred_final

results_df['absolute_error'] = abs(
    results_df['actual_traffic'] - results_df['predicted_traffic']
)

results_df['residual'] = (
    results_df['actual_traffic'] - results_df['predicted_traffic']
)

results_df.head()
```

```python
def traffic_period(hour):
    
    if 0 <= hour < 6:
        return 'Night'
    
    elif 6 <= hour < 10:
        return 'Morning Peak'
    
    elif 10 <= hour < 16:
        return 'Daytime'
    
    elif 16 <= hour < 19:
        return 'Evening Peak'
    
    else:
        return 'Night'


results_df['traffic_period'] = results_df['hour'].apply(traffic_period)
```

```python
period_performance = (
    results_df
    .groupby('traffic_period')
    .agg(
        observations=('actual_traffic', 'count'),
        MAE=('absolute_error', 'mean'),
        actual_traffic=('actual_traffic', 'mean'),
        predicted_traffic=('predicted_traffic', 'mean')
    )
    .round(2)
)

# Arrange in logical order
period_performance = period_performance.reindex([
    'Night',
    'Morning Peak',
    'Daytime',
    'Evening Peak'
])

period_performance
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

period_performance['MAE'].plot(
    kind='bar'
)

plt.title('Model Prediction Error by Traffic Period')
plt.xlabel('Traffic Period')
plt.ylabel('Mean Absolute Error (Vehicles)')
plt.xticks(rotation=0)

plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Add day type

results_df['day_type'] = results_df['is_weekend'].map({
    0: 'Weekday',
    1: 'Weekend'
})

# Calculate model performance by day type

daytype_performance = (
    results_df
    .groupby('day_type')
    .agg(
        observations=('actual_traffic', 'count'),
        MAE=('absolute_error', 'mean'),
        RMSE=('residual', lambda x: (x**2).mean()**0.5),
        avg_actual_traffic=('actual_traffic', 'mean')
    )
    .round(2)
)

daytype_performance
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

daytype_performance['MAE'].plot(kind='bar')

plt.title('Model Prediction Error: Weekday vs Weekend')
plt.xlabel('Day Type')
plt.ylabel('Mean Absolute Error (Vehicles)')
plt.xticks(rotation=0)

plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
